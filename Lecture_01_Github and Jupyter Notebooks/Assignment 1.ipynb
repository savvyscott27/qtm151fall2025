```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# Create figure
fig, ax = plt.subplots(figsize=(12, 8))
ax.axis("off")

# Nodes (rectangles for decisions/outcomes)
nodes = {
    "start": (0.5, 0.95, "What is your research question?"),
    "cause": (0.2, 0.75, "Do you want to test\nif one variable causes another?"),
    "relationship": (0.5, 0.75, "Do you want to see\nif two variables are related?"),
    "description": (0.8, 0.75, "Do you want to describe\nbehavior or phenomena?"),
    "experimental": (0.2, 0.55, "→ Use Experimental Design\n(Cause & Effect)"),
    "correlational": (0.5, 0.55, "→ Use Correlational Design\n(Relationships, no causality)"),
    "descriptive": (0.8, 0.55, "→ Use Descriptive Design\n(Case study, survey, observation)"),
    "multi": (0.5, 0.3, "Can you combine methods?\n→ Multi-Method Approach\n(e.g., survey + experiment)")
}

# Draw rectangles for each node
for key, (x, y, text) in nodes.items():
    ax.text(x, y, text, ha="center", va="center",
            bbox=dict(boxstyle="round,pad=0.5", facecolor="#e6f2ff", edgecolor="black"),
            fontsize=11)

# Draw arrows
def draw_arrow(start, end):
    ax.annotate("",
                xy=(nodes[end][0], nodes[end][1]+0.05),
                xytext=(nodes[start][0], nodes[start][1]-0.05),
                arrowprops=dict(arrowstyle="->", lw=2))

draw_arrow("start", "cause")
draw_arrow("start", "relationship")
draw_arrow("start", "description")
draw_arrow("cause", "experimental")
draw_arrow("relationship", "correlational")
draw_arrow("description", "descriptive")
draw_arrow("experimental", "multi")
draw_arrow("correlational", "multi")
draw_arrow("descriptive", "multi")

plt.title("Choosing a Research Design in Psychology", fontsize=14, pad=20)
plt.show()
```
